## 5. Zusammenführen von String-Spalten

Derzeit sind das Datum und die Uhrzeit jedes Polizeistopps in zwei separaten Objektspalten gespeichert: `stop_date` und `stop_time`.

Sie werden diese beiden Spalten zu einer einzigen Spalte kombinieren und anschließend in das Datumsformat (`datetime`) umwandeln.  
Dies ermöglicht den einfachen Zugriff auf datumsbezogene Attribute, die wir später im Projekt verwenden werden.

**ANWEISUNGEN**

- Verwenden Sie eine String-Methode, um `stop_date` und `stop_time` (getrennt durch ein Leerzeichen) zu verketten, und speichern Sie das Ergebnis in einer neuen Variable namens `combined`.
- Wandeln Sie `combined` in das Datumsformat (`datetime`) um und speichern Sie das Ergebnis in einer neuen Spalte namens `stop_datetime`.
- Untersuchen Sie die `.dtypes` des DataFrames, um zu bestätigen, dass `stop_datetime` nun den Typ `datetime` hat.


### Ich kombiniere die relevanten Spalten, um die Daten zu vereinfachen und den Fokus auf das Wesentliche der Analyse zu richten.


Jetzt werden wir aus den Spalten `stop_date` und `stop_time` eine einzelne Spalte erstellen und diese anschließend in das `datetime`-Format umwandeln.

```python
ri.head()
```

Ich werde zwei Pandas-Serien mit der Funktion `str.cat()` zusammenführen.  
Dabei übergebe ich `stop_date` an `.str.cat()` und verbinde es mit `ri.stop_time`.  
Das Ergebnis speichere ich in der Variablen `combined`.


```python
# stop_date und stop_time verbinden
combined = ri.stop_date.str.cat(ri.stop_time, sep=" ")

# neue Serie mit Datum und Uhrzeit
combined
```

Jetzt erstelle ich die Spalte `stop_datetime`, der ich die Serie `combined` zuweise.  
Da sie sowohl Datum als auch Uhrzeit enthält, nenne ich sie `stop_datetime`.  
Ich konvertiere sie mit `to_datetime()` in den `datetime`-Typ, um anschließend Datumsfunktionen verwenden zu können.

```python
# Neue datetime-Spalte erstellen
ri["stop_datetime"] = pd.to_datetime(combined)
```

```python
# Alte Spalten entfernen
ri.drop(["stop_date", "stop_time"], axis="columns", inplace=True)
```
