Werden **males** möglicherweise öfter gefilzt als **females**, weil sie von Polizeibeamten als risikoreicher eingeschätzt werden?

```python
ri.search_conducted

# Das ist eine Series mit True- und False-Werten.
# Wenn man sie auf ``ri`` anwendet, erhält man ein ``DataFrame`` mit den Zeilen, bei denen der Wert ``True`` ist.
```
